# Ollama 대화형 챗봇

Ollama가 실행 중이고 `llama3.2:3b` 모델이 준비되어 있어야 합니다. 터미널에서 `ollama pull llama3.2:3b`로 모델을 받으세요.

아래 셀을 순서대로 실행합니다. 대화 종료는 `exit`, `quit`, `q`이며, 대화 셀을 다시 실행하면 기록이 초기화됩니다.

```python
%pip install langchain-ollama
```

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

llm = ChatOllama(model="llama3.2:3b")
```

```python
messages = [SystemMessage("You are a helpful assistant.")]

while True:
    user_input = input("메시지를 입력하세요: ").strip()

    if user_input.lower() in ["exit", "quit", "q"]:
        print("Goodbye!")
        break

    if not user_input:
        continue

    # 노트북 출력 영역에도 사용자 메시지를 남깁니다.
    print(f"You\t: {user_input}")
    messages.append(HumanMessage(content=user_input))

    response = llm.invoke(messages)
    print(f"Bot\t: {response.content}\n")
    messages.append(AIMessage(content=response.content))
```

대화 종료 후 아래 셀을 실행하면 시스템 메시지를 제외한 전체 대화 기록을 다시 볼 수 있습니다.

```python
for message in messages:
    if isinstance(message, HumanMessage):
        print(f"You\t: {message.content}")
    elif isinstance(message, AIMessage):
        print(f"Bot\t: {message.content}\n")
```
