```python
import pymupdf
import os

# PDF 파일 경로
pdf_file_path = "data/From Passive Retrieval to Active Memory Navigation Learning to Use Memory as a Structured Action Space.pdf"

# PDF 열기
doc = pymupdf.open(pdf_file_path)

# 헤더 / 푸터 높이
header_height = 80
footer_height = 80

full_text = ""

# 페이지별 텍스트 추출
for page in doc:
    rect = page.rect  # 페이지 크기

    # 헤더 영역
    header = page.get_text(
        clip=(0, 0, rect.width, header_height)
    )

    # 푸터 영역
    footer = page.get_text(
        clip=(
            0,
            rect.height - footer_height,
            rect.width,
            rect.height
        )
    )

    # 본문 영역
    text = page.get_text(
        clip=(
            0,
            header_height,
            rect.width,
            rect.height - footer_height
        )
    )

    full_text += text + "\n------------------------------------\n"


# PDF 파일명만 추출
pdf_file_name = os.path.basename(pdf_file_path)

# 확장자 제거
pdf_file_name = os.path.splitext(pdf_file_name)[0]


# 출력 폴더 생성
output_dir = "output"
os.makedirs(output_dir, exist_ok=True)


# TXT 파일 경로
txt_file_path = f"{output_dir}/{pdf_file_name}_with_preprocessing.txt"


# TXT 파일 저장
with open(txt_file_path, "w", encoding="utf-8") as f:
    f.write(full_text)


print(f"텍스트 추출 완료: {txt_file_path}")
```
