````python
from dotenv import load_dotenv
from openai import OpenAI
from ollama import chat

load_dotenv()

client = OpenAI()
````

## url 이미지 분석 요청

````python
# 이미지 URL
image_url = (
    "https://images.unsplash.com/photo-1786334633564-81b58c02e5bf?q=80&w=1748"
)


# 메시지 구성
messages = [
    {
        "role": "user",
        "content": [
            {
                "type": "text",
                "text": "이 이미지에 대해 설명해 주세요."
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": image_url
                }
            }
        ]
    }
]


# OpenAI API 호출
response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=messages
)

# 응답 확인
print(response.choices[0].message.content)
````

## 로컬 이미지 base 64로 변환

````python
import base64


def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(
            image_file.read()
        ).decode("utf-8")


image_path = "img/alex-ramon-aY_nzJTardo-unsplash.jpg"


# 이미지를 Base64로 인코딩
base64_image = encode_image(image_path)

messages = [
    {
        "role": "user",
        "content": [
            {
                "type": "text",
                "text": "이 이미지에 대해 설명해주세요."
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": f"data:image/jpeg;base64,{base64_image}"
                }
            }
        ]
    }
]


response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=messages
)


print(response.choices[0].message.content)
````

## 두 이미지 비교
- gpt와 ollama 모델 비교

````python
# 두 이미지 Base64 인코딩

mountain01_base64 = encode_image(
    "img/alex-ramon-aY_nzJTardo-unsplash.jpg"
)

mountain02_base64 = encode_image(
    "img/hunter-owens-FoH8Lt_H2Tc-unsplash.jpg"
)


mountain01_path = "img/oecd_rnd_2021_large.png"
mountain02_path = "img/oecd_rnd_2022_large.png"
````

````python
# gpt

# 메시지 구성
messages = [
    {
        "role": "user",
        "content": [
            {
                "type": "text",
                "text": "두 산악 풍경의 차이점을 설명해 주세요."
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": (
                        f"data:image/jpeg;base64,"
                        f"{mountain01_base64}"
                    )
                }
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": (
                        f"data:image/jpeg;base64,"
                        f"{mountain02_base64}"
                    )
                }
            }
        ]
    }
]


# OpenAI API 호출
response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=messages
)


# AI 응답 출력
print(response.choices[0].message.content)
````

````python
# ollama
response = chat(
    model="llava:latest",
    messages=[
        {
            "role": "user",
            "content": "Please explain the differences between these two mountain landscapes in detail.",
            "images": [
                mountain01_base64,
                mountain02_base64,
            ],
        }
    ],
)

print(response.message.content)
````

해석:
제공하신 이미지는 각각 고유한 특징을 지닌 두 개의 서로 다른 산악 지형을 보여줍니다.

1. 산악 지형 A: 이 지형은 주로 소나무와 전나무로 이루어진 울창한 상록수림이 산의 낮은 경사를 덮고 있는 것이 특징입니다. 지형은 거칠고 바위투성이로, 상업적 목적으로 크게 개간되거나 개발되지 않은 지역임을 나타냅니다. 이 풍경에는 보이는 구조물이 없어, 이곳이 황야나 외진 지역일 가능성을 시사합니다.

2. 산악 풍경 B: 이 풍경은 더 발달되고 인간의 영향을 많이 받은 환경을 특징으로 합니다. 전경에는 산악 지역을 찾는 방문객들을 위한 휴가용 주택이나 오두막으로 보이는 건물들이 있는 작은 마을이 있습니다. 식생이 드물어 배경에 있는 산들이 선명하게 보인다. 주변 지역에도 몇 그루의 나무가 흩어져 있으며, 전체적인 경관은 조경 A에 보인 것보다 덜 빽빽해 보입니다.

두 지형의 주요 차이점은 다음과 같습니다:

- 식생 밀도: 경관 A는 식생이 훨씬 더 조밀하게 덮여 있어, 인간 활동으로 일부 산림 지역이 개간된 경관 B에 비해 더 외지고 손대지 않은 상태임을 나타냅니다.
  
- 인간 인프라의 존재: 경관 B는 건물 형태로 인간의 존재를 보여주며, 더 발달했거나 방문된 지역임을 시사합니다. 반면에, 풍경 A에는 눈에 띄는 구조물이 없어 그 야생의 특성을 더욱 강조합니다.
   
- 식생 다양성: 나무의 종류는 비슷해 보이지만, 햇빛 노출, 토양의 질, 고도 차이로 인해 약간의 차이가 있을 수 있습니다. 하지만 가장 큰 차이점은 Landscape B가 나무 분포가 더 드문드문해서 경관을 더 잘 볼 수 있다는 점입니다.
    
이 두 풍경은 청정한 야생지와 인간의 영향을 받은 환경 사이에 뚜렷한 대조를 이루며, 방문객이나 주민 모두에게 각기 다른 특별한 경험을 선사합니다.

````python
# ollama
response = chat(
    model="gemma4:e4b",
    messages=[
        {
            "role": "user",
            "content": "두 산악 풍경의 차이점을 설명해 주세요.",
            "images": [
                mountain01_path,
                mountain02_path,
            ],
        }
    ],
)

print(response.message.content)
````

## 차트 비교를 잘 해주는지 확인
- gpt와 ollama 비교

````python
oecd_rnd_2021_base64 = encode_image(
    "img/oecd_rnd_2021_large.png"
)

oecd_rnd_2022_base64 = encode_image(
    "img/oecd_rnd_2022_large.png"
)

# ollama는 encode를 할 필요 없다
# ollama Python 라이브러리에서는 로컬 이미지 파일을 images에 경로로 직접 넣을 수 있다
oecd_rnd_2021_path = "img/oecd_rnd_2021_large.png"
oecd_rnd_2022_path = "img/oecd_rnd_2022_large.png"
````

````python

messages = [
    {
        "role": "user",
        "content": [
            {
                "type": "text",
                "text": (
                    "첫 번째는 2021년 데이터이고, "
                    "두 번째는 2022년 데이터입니다. "
                    "이 데이터에 대해 설명해 주세요. "
                    "어떤 변화가 있었나요? "
                    "한국 중심으로 설명해 주세요."
                )
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": (
                        f"data:image/png;base64,"
                        f"{oecd_rnd_2021_base64}"
                    )
                }
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": (
                        f"data:image/png;base64,"
                        f"{oecd_rnd_2022_base64}"
                    )
                }
            }
        ]
    }
]


response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=messages
)


print(response.choices[0].message.content)
````

````python
prompt_text = """The first image is 2021 data, and the second is 2022 data. 
Please explain this data. 
What changes have occurred? 
Please explain it focusing on South Korea."""

response = chat(
    model="llava:latest",
    messages=[
        {
            "role": "user",
            "content": prompt_text, 
            "images": [
                oecd_rnd_2021_path,
                oecd_rnd_2022_path,
            ],
        }
    ],
)

print(response['message']['content'])
````

해석:
이 이미지는 대한민국의 2021년과 2022년 사이 특정 데이터 포인트를 비교한 두 개의 별도 막대 차트를 보여줍니다. 차트는 기술, 보건, 교육 및 기타 분야와 관련된 다양한 통계를 보여줍니다. 변경 사항을 정리하면 다음과 같습니다:

1. **보건 및 의료 기술**: 보건 분야와 의료 기술 분야에 종사하는 사람 수를 나타내는 두 개의 막대가 있습니다. 2021년에는 각 범주에 약 500명을 조금 넘는 인원이 있었습니다. 2022년까지 이 두 수치는 모두 두 배로 늘어나, 이 분야에서 약 1,600명이 일하고 있습니다.

2. **교육**: 교육 분야 역시 상당한 성장을 경험했습니다. 2021년에는 약 390명이었고, 2022년에는 그 수가 두 배 이상 늘어나 850명을 넘었습니다. 이는 교육 활동이나 노동력이 크게 증가했음을 시사합니다.

3. **통신**: 통신 부문은 2021년 약 400명에서 2022년 약 370명으로 소폭 감소했습니다.

4. **기술 및 소프트웨어**: 기술 및 소프트웨어 분야의 인원 수는 2021년 약 350명에서 2022년 약 630명으로 크게 증가해 이 분야의 상당한 성장을 보여줍니다.

5. **기타**: "기타"라는 카테고리도 증가를 보이지만, 다른 부문에 비해 증가 폭은 크지 않습니다.

전반적으로 데이터는 2021년부터 2022년 사이에 한국의 보건 및 의료 기술, 교육, 그리고 기술 및 소프트웨어 분야에서 눈에 띄는 확장이 있었음을 시사합니다. 이들 분야의 성장은 혁신, 의료의 현대화, 그리고 교육 시스템 내 디지털 전환에 대한 강조를 반영하는 것일 수 있습니다.

````python
# ollama
prompt_text = """첫 번째는 2021년 데이터이고,
두 번째는 2022년 데이터입니다.
이 데이터에 대해 설명해 주세요.
어떤 변화가 있었나요? 한국 중심으로 설명해 주세요."""

response = chat(
    model="gemma4:e4b",
    messages=[
        {
            "role": "user",
            "content": prompt_text,
            "images": [
                oecd_rnd_2021_path,
                oecd_rnd_2022_path,
            ],
        }
    ],
)

print(response.message.content)
````

(화산암 vs. 퇴적암 등)#### 2. 생태적 특징 및 식생 (Ecology & Vegetation)* **식물군계:** 온대림(나무가 울창한지), 아고산대 초원(풀이나 관목 위주인지), 혹은 건조하고 메마른 황무지 형태의 식생이 우세한지?* **색감/활력도:** 녹음이 풍부하여 생명력이 느껴지는지, 아니면 흙빛과 돌빛이 주를 이루어 황량하게 보이는지?#### 3. 기후적 분위기 (Atmosphere & Climate)* **날씨:** 안개나 구름이 자주 끼는 습한 느낌인지, 아니면 맑고 청명하며 건조한 느낌인지?* **분위기/톤:** 웅장하고 장엄하여 압도적인 느낌(Dramatic)을 주는지, 혹은 고요하고 평화로운 (Peaceful) 분위기를 풍기는지?#### 4. 인간의 영향 (Human Influence)* **개발 정도:** 인공적인 구조물(도로, 마을, 건물 등)이 많이 보이나요? 아니면 완전히 자연 그대로의 모습(Wilderness)을 유지하고 있나요?---사진을 다시 올려주시면 이 기준들을 바탕으로 **'어떤 풍경은 ~한 특징을 가진 반면, 다른 풍경은 ~라는 대비되는 특성'**과 같이 자세하게 설명해 드리겠습니다. 😊
