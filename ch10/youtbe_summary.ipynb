```python
from youtube_search import YoutubeSearch

videos = YoutubeSearch("한 눈에 보는 대규모 언어 모델", max_results=5).to_dict()
videos
```

### 유튜브 링크 추출

```python
video_url = 'http://youtube.com' + videos[1]['url_suffix']
video_url
```

### 유튜브 자막 가져오기

```python
from langchain_community.document_loaders import YoutubeLoader

loader = YoutubeLoader.from_youtube_url(
    video_url, 
    language=['ko', 'en'] # 자막 언어
)

loader.load()
```

```python
for v in videos:
    # url_suffix를 이용하여 video_url을 만듭니다.
    v['video_url'] = 'https://youtube.com' + v['url_suffix']
    
    # YoutubeLoader를 이용하여 비디오를 로드합니다.
    loader = YoutubeLoader.from_youtube_url(
        v['video_url'],
        language=['ko', 'en']
    )

    v['content'] = loader.load()

videos
```

### 길이가 10분 이하인 영상만 남기기

```python
print('총 영상 수:', len(videos))

# 특정 유튜버의 영상만 남깁니다.
videos = [
    v for v in videos
    if v['channel'] == '3Blue1Brown'
]

# 영상 길이가 10분 이하인 영상만 남깁니다.
videos = [
    v for v in videos
    if len(v['duration'].split(':')) == 2
    and int(v['duration'].split(':')[0]) <= 10
]

print('특정 채널 + 10분 이하 영상 수:', len(videos))

videos[0]
```

```python
# 랭체인, openai 임포트
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-5.6-luna")
model.invoke('안녕?') # 언어 모델이 잘 설정되었는지 테스트
```

```python
from langchain_core.prompts import ChatPromptTemplate

# 동영상 요약 프롬프트 작성
prompt = ChatPromptTemplate.from_messages([
    ("system", "다음 영상에 대한 요약을 한국어로 만들어줘:\n\n{content}")
])

chain = prompt | model

result = chain.invoke({"content": videos[0]["content"]})
print(result.content)
```

```python
# 모든 비디오에 대해 요약을 생성
from tqdm import tqdm  # tqdm은 진행 상황을 보여주는 라이브러리

for v in tqdm(videos):
    v['summary'] = chain.invoke({"content": v['content']}).content

videos
```
