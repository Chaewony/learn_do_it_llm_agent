```python
from ollama import chat
from pprint import pprint
```

```python
stream = chat(
    model="gemma4:e4b",
    messages=[
        {
            "role": "user",
            "content": "2022년 월드컵 우승 팀은 어디야?"
        }
    ],
    stream=True,
)
```

```python
for chunk in stream:
    pprint(chunk.model_dump())
    break
```

```python
for chunk in stream:
    print(chunk.message.content, end="", flush=True)
```
