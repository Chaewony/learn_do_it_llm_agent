```python
from dotenv import load_dotenv
from openai import OpenAI
from pprint import pprint

load_dotenv()

client = OpenAI()
```

```python
response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant."
        },
        {
            "role": "user",
            "content": "2022년 월드컵 우승 팀은 어디야?"
        }
    ]
)
```

```python
data = response.model_dump()
pprint(data)
```

```python
print("응답 ID:")
print(data["id"])

print("\n사용 모델:")
print(data["model"])

print("\n답변:")
print(data["choices"][0]["message"]["content"])

print("\n토큰 사용량:")
pprint(data["usage"])
```

## 호출 - 응답 구조

OpenAI API
│
├── 요청(Request)
│   ├── model
│   ├── temperature
│   └── messages
│       ├── system
│       └── user
│
└── 응답(Response)
    │
    ├── id
    │
    ├── model
    │
    ├── choices
    │   └── message
    │       ├── role
    │       └── content
    │
    └── usage
        ├── prompt_tokens
        ├── completion_tokens
        └── total_tokens
