```python
# 공통
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

client = OpenAI() # OpenAI()는 기본적으로 환경변수 OPENAI_API_KEY를 찾아서 사용합니다.
```

```python
# mirror
response_mirror = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {
            "role": "system",
            "content": "너는 백설공주 이야기 속의 마법 거울이야. 그 이야기의 캐릭터에 부합하게 답변해줘."
        },
        {
            "role": "user",
            "content": "세상에서 누가 제일 아름답니?"
        },
    ]
)

print(response_mirror.choices[0].message.content)  # response의 내용만 출력
```

```python
# joker
response_joker = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {
            "role": "system",
            "content": "너는 배트맨에 나오는 조커야. 조커의 악당 캐릭터에 맞게 답변 해 줘."
        },
        {
            "role": "user",
            "content": "세상에서 누가 제일 아름답니?"
        },
    ]
)

print(response_joker.choices[0].message.content)  # response의 내용만 출력
```

```python
# no prompting
response_no_prompting = client.chat.completions.create(
    model="gpt-4o",
    temperature=0.9,
    messages=[
        {
            "role": "system",
            "content": "너는 유치원생이야. 유치원생처럼 답변해 줘."
        },
        {
            "role": "user",
            "content": "오리"
        },
    ]
)

print(response_no_prompting.choices[0].message.content)  # response의 내용만 출력
```

```python
# one shot prompting
response_one_shot_prompting = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {
            "role": "system",
            "content": "너는 유치원생이야. 유치원생처럼 답변해 줘."
        },
        {
            "role": "user",
            "content": "참새"
        },
        {
            "role": "assistant",
            "content": "짹짹"
        },
        {
            "role": "user",
            "content": "오리"
        },
    ]
)

print(response_one_shot_prompting.choices[0].message.content)  # response의 내용만 출력
```
