```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-5.6-luna")
```

```python
from langchain_core.messages import HumanMessage

model.invoke([
    HumanMessage(content="안녕? 나는 이성용이야.")
])
```

```python
model.invoke([
    HumanMessage(content="내 이름이 뭐지?")
])
```

### Q. 어떻게 이름을 기억하는 것 처럼 보이는가?
- invoke() 자체가 기억하는 게 아니라, LangChain의 Runnable이 입력으로 받은 messages를 그대로 모델에 전달하기 때문에 이전 대화가 유지되는 것
- invoke()는 대화를 기억하는 기능이 아니라, 전달받은 대화 기록을 포함한 입력을 실행하는 역할
